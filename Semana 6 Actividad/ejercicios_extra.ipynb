### Ejercicio 1

```python
base = float(input("Ingresa la base: "))
altura = float(input("Ingresa la altura: "))

area = (base * altura) / 2

print(f"El área del triángulo es: {area}")
```

### Ejercicio 9

```python
a = int(input("Primer número: "))
b = int(input("Segundo número: "))
c = int(input("Tercer número: "))

if a >= b and a >= c:
    mayor = a
elif b >= a and b >= c:
    mayor = b
else:
    mayor = c

print(f"El mayor de los tres números es: {mayor}")
```

### Ejercicio 10

```python
suma = 0
numero = int(input("Ingresa un número (0 para terminar): "))

while numero != 0:
    suma += numero
    numero = int(input("Ingresa un número (0 para terminar): "))

print(f"La suma de los números es: {suma}")
```

### Ejercicio 12

```python
suma = 0

for i in range(1, 21):
    if i % 2 != 0:
        continue
    suma += i

print(f"La suma de los números pares del 1 al 20 es: {suma}")
```

### Ejercicio 13

```python
def sumar_lista(lista):
    total = 0
    for elemento in lista:
        total += elemento
    return total

numeros = [10, 20, 30, 40]
print(f"La suma de los elementos es: {sumar_lista(numeros)}")
```

### Ejercicio 14

```python
def maximo_lista(lista):
    mayor = lista[0]
    for elemento in lista:
        if elemento > mayor:
            mayor = elemento
    return mayor

numeros = [7, 3, 9, 2]
print(f"El elemento máximo es: {maximo_lista(numeros)}")
```

### Ejercicio 15

```python
matriz = [[1, 2, 3],
          [4, 5, 6],
          [7, 8, 9]]

suma = 0
for i in range(len(matriz)):
    suma += matriz[i][i]

print(f"La suma de la diagonal principal es: {suma}")
```

### Ejercicio 16

```python
def mostrar_promedio(lista):
    total = 0
    for elemento in lista:
        total += elemento
    promedio = total / len(lista)
    print(f"El promedio es: {promedio:.2f}")

mostrar_promedio([8, 9, 10, 7])
```

### Ejercicio 18

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

numero = int(input("Ingresa un número: "))
print(f"El factorial de {numero} es: {factorial(numero)}")
```

### Ejercicio 19

```python
def sumar_tupla(tupla):
    total = 0
    for elemento in tupla:
        total += elemento
    return total

datos = (4, 8, 15, 16, 23, 42)

print(f"El tercer elemento es: {datos[2]}")
print(f"La suma de la tupla es: {sumar_tupla(datos)}")
```
